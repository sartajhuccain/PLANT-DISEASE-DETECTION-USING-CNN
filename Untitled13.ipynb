```python
import os
import json
import random
import shutil
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
DATASET_ID = "vipoooool/new-plant-diseases-dataset"

DOWNLOAD_DIR = "/content/kaggle_download"
EXTRACT_DIR = "/content/plant_dataset"

os.makedirs(DOWNLOAD_DIR, exist_ok=True)
os.makedirs(EXTRACT_DIR, exist_ok=True)

!kaggle datasets download -d vipoooool/new-plant-diseases-dataset -p /content/kaggle_download
```

```python
zip_files = [
    os.path.join(DOWNLOAD_DIR, name)
    for name in os.listdir(DOWNLOAD_DIR)
    if name.lower().endswith(".zip")
]

if not zip_files:
    raise FileNotFoundError(
        "ZIP file nahi mila. Kaggle download aur API permissions check karein."
    )

for zip_path in zip_files:
    print("Extracting:", os.path.basename(zip_path))

    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(EXTRACT_DIR)

print("Extraction complete.")
print("Dataset path:", EXTRACT_DIR)
print("Top-level contents:", os.listdir(EXTRACT_DIR)[:20])
```

```python
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def image_count(folder):
    if not os.path.isdir(folder):
        return 0

    count = 0
    for root, _, files_list in os.walk(folder):
        count += sum(
            filename.lower().endswith(IMAGE_EXTENSIONS)
            for filename in files_list
        )

    return count


def has_class_subfolders(folder):
    if not os.path.isdir(folder):
        return False

    for name in os.listdir(folder):
        subfolder = os.path.join(folder, name)

        if os.path.isdir(subfolder) and image_count(subfolder) > 0:
            return True

    return False


TRAIN_DIR = None
VALID_DIR = None
TEST_DIR = None

for root, dirs, _ in os.walk(EXTRACT_DIR):
    dirs.sort()
    folder_name = os.path.basename(root).lower()

    if not has_class_subfolders(root):
        continue

    if folder_name in ("train", "training"):
        TRAIN_DIR = root
    elif folder_name in ("valid", "validation", "val"):
        VALID_DIR = root
    elif folder_name in ("test", "testing"):
        TEST_DIR = root

print("TRAIN_DIR:", TRAIN_DIR)
print("VALID_DIR:", VALID_DIR)
print("TEST_DIR:", TEST_DIR)

if TRAIN_DIR is None:
    raise FileNotFoundError(
        "Training folder nahi mila. Extracted folders ka output check karein."
    )

print("Training images:", image_count(TRAIN_DIR))

if VALID_DIR:
    print("Validation images:", image_count(VALID_DIR))

if TEST_DIR:
    print("Test images:", image_count(TEST_DIR))
```

```python
def get_class_names(folder):
    return sorted([
        name
        for name in os.listdir(folder)
        if os.path.isdir(os.path.join(folder, name))
        and image_count(os.path.join(folder, name)) > 0
    ])

class_names = get_class_names(TRAIN_DIR)
NUM_CLASSES = len(class_names)

print("Number of classes:", NUM_CLASSES)

for name in class_names:
    print(name)

if NUM_CLASSES < 2:
    raise ValueError(
        "At least 2 disease class folders required. Check TRAIN_DIR."
    )
```

```python
IMG_SIZE = (160, 160)
BATCH_SIZE = 32

if VALID_DIR is not None:
    valid_class_names = get_class_names(VALID_DIR)

    if valid_class_names != class_names:
        raise ValueError(
            "Training aur validation class folders match nahi karte."
        )

    train_ds = tf.keras.utils.image_dataset_from_directory(
        TRAIN_DIR,
        class_names=class_names,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="int",
        shuffle=True,
        seed=SEED
    )

    val_ds = tf.keras.utils.image_dataset_from_directory(
        VALID_DIR,
        class_names=class_names,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="int",
        shuffle=False
    )

else:
    train_ds = tf.keras.utils.image_dataset_from_directory(
        TRAIN_DIR,
        class_names=class_names,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="int",
        validation_split=0.2,
        subset="training",
        seed=SEED
    )

    val_ds = tf.keras.utils.image_dataset_from_directory(
        TRAIN_DIR,
        class_names=class_names,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="int",
        validation_split=0.2,
        subset="validation",
        seed=SEED
    )

train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)

print("Training and validation datasets are ready.")
```

```python
plt.figure(figsize=(12, 8))

for images, labels in train_ds.take(1):
    count = min(9, images.shape[0])

    for i in range(count):
        plt.subplot(3, 3, i + 1)
        plt.imshow(tf.cast(images[i], tf.uint8).numpy())
        plt.title(class_names[int(labels[i].numpy())])
        plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from tensorflow.keras import layers, models, regularizers

data_augmentation = models.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomZoom(0.10),
    layers.RandomContrast(0.10)
], name="data_augmentation")

model = models.Sequential([
    layers.Input(shape=(160, 160, 3)),

    data_augmentation,
    layers.Rescaling(1.0 / 255),

    # Convolution block 1
    layers.Conv2D(32, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    # Convolution block 2
    layers.Conv2D(64, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    # Convolution block 3
    layers.Conv2D(128, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    # Convolution block 4
    layers.Conv2D(256, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),

    # Classification head
    layers.GlobalAveragePooling2D(),
    layers.Dense(
        256,
        activation="relu",
        kernel_regularizer=regularizers.l2(0.001)
    ),
    layers.Dropout(0.45),
    layers.Dense(NUM_CLASSES, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("CNN compiled successfully.")
```

```python
BEST_MODEL_PATH = "/content/best_plant_disease.keras"

callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        BEST_MODEL_PATH,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-7,
        verbose=1
    )
]

print("Callbacks are ready.")
```

```python
EPOCHS = 5

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)
```

```python
 plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("CNN Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("CNN Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
best_model = tf.keras.models.load_model(BEST_MODEL_PATH)

val_loss, val_accuracy = best_model.evaluate(
    val_ds,
    verbose=1
)

print(f"Validation Accuracy: {val_accuracy * 100:.2f}%")
print(f"Validation Loss: {val_loss:.4f}")
```

```python
from sklearn.metrics import classification_report

y_true = []
y_pred = []

for images, labels in val_ds:
    probabilities = best_model.predict(images, verbose=0)

    y_true.extend(labels.numpy().astype(int).tolist())
    y_pred.extend(np.argmax(probabilities, axis=1).tolist())

print(classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    target_names=class_names,
    zero_division=0,
    digits=4
))
```
